### Задание 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

Решение:

а) Мы выбираем из 10 фильмов 5 номинантов, которые получат разные призы, то есть нам важно при рассмотрении вариантов <u>и порядок и состав</u>, так что берем размещения. Но один фильм может выиграть несколько номинаций, так что используем <u>размещения с повторениями</u>:
$$
\tilde{A}^{5}_{10} = 10^{5}
$$

```python
import scipy.special as sc
import itertools as it
```

```python
source_list = [i for i in range(1,11)]

print(len(set(it.product(source_list, repeat=5))))
print(10**5)
```

б) Мы выбираем из 10 фильмов 5 номинантов, которые получат одинаковые призы, то есть нам важно при рассмотрении вариантов <u>только состав </u>, так что берем сочетания. Но один фильм может выиграть несколько номинаций, так что используем <u>сочетание с повторениями</u>:
$$
\tilde{C}^{5}_{10} = {C}^{5}_{10 + 5 - 1} = {C}^{5}_{14} = \frac{14!}{5! \cdot (14 - 5)!} = \frac{14!}{5! \cdot 9!}
$$

```python
print(len(set(it.combinations_with_replacement(source_list, r=5))))
print(sc.factorial(10 + 5 - 1) / (sc.factorial(5) * sc.factorial(10 - 1)))
print(sc.comb(14, 5))
```

### Задание 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

Решение:
В сумме у нас всего семь цифр, так что будем рассматривать их перестановки:
$$
P_{7}(3, 2, 2) = \frac{7!}{3! \cdot 2! \cdot 2!}
$$

```python
source_list = [4, 4, 4, 5, 5, 6, 6]
len_list = len(source_list)

print(len(set(it.permutations(source_list, len_list))))
print(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2)))
```

### Задание 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

Решение:

a) Рассмотрим событие $A$ – получится слово "ТОР"
$$
P(A) = \frac{1}{A^{3}_{6}}
$$

```python
from sympy import Rational

print(Rational(1, sc.perm(6, 3)))
print(round(1 / sc.perm(6, 3), 4))
```

б) Рассмотрим событие $B$ – получится слово "ТЕОРИЯ"
$$
P(B) = \frac{1}{P_{6}} = \frac{1}{6!}
$$

```python
print(Rational(1, sc.factorial(6)))
print(round(1 / sc.factorial(6), 6))
```

### Задание 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Решение:

Рассмотрим событие $A$ – получится слово "АНАНАС"
$$
P(A) = \frac{1}{P_{6}(3, 2, 1)} = \frac{3! \cdot 2! \cdot 1!}{6!} = \frac{3! \cdot 2!}{6!}
$$

```python
print(Rational(sc.factorial(3) * sc.factorial(2), sc.factorial(6)))
print(round((sc.factorial(3) * sc.factorial(2)) / sc.factorial(6), 4))
```

```python
print(Rational(1, len(set(it.permutations("АААННС")))))
print(round(1 / len(set(it.permutations("АААННС"))), 4))
```

### Задание 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

Решение:

a) Рассмотрим событие $A$ – никакие два лица одного пола не сядут рядом
$$
P(A) = \frac{2 \cdot (P_{5})^2}{P_{10}} = \frac{2 \cdot (5!)^2}{10!}
$$

```python
print(round(2 * sc.factorial(5)**2 / sc.factorial(10), 5))
```

a) Рассмотрим событие $B$ – все супружеские пары сядут рядом
$$
P(B) = \frac{2^5 \cdot P_{5}}{P_{10}} = \frac{2^5 \cdot 5!}{10!}
$$

```python
print(round(2**5 * sc.factorial(5) / sc.factorial(10), 5))
```

### Задание 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Решение:

a) Рассмотрим событие $A$ – 7 пассажиров уместилось в двух купе
$$
P(A) = \frac{C^{2}_{9} \cdot C^{7}_{8}}{C^{7}_{36}}
$$

```python
print(round((sc.comb(9, 2) * sc.comb(8, 7)) / sc.comb(36, 7), 7))
```

б) Рассмотрим событие $B$ – 7 пассажиров уместилось в семи купе
$$
P(B) = \frac{4^7 \cdot C^{7}_{9}}{C^{7}_{36}}
$$

```python
print(round((4**7 * sc.comb(9, 7)) / sc.comb(36, 7), 5))
```

в) Рассмотрим событие $C$ – 7 пассажиров уместилось в трех купе
$$
P(C) = \frac{C^{3}_{9} \cdot (6C^{4}_{4}C^{2}_{4}C^{1}_{4} + 3C^{3}_{4}C^{3}_{4}C^{1}_{4} + 3C^{3}_{4}C^{2}_{4}C^{2}_{4})}{C^{7}_{36}}
$$

```python
n_1 = sc.comb(9, 3) * (6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + 
                       3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + 
                       3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2))
print(round(n_1 / sc.comb(36, 7), 5))
```
